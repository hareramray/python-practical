## 1. Use the standard library

A module is a Python file or another importable unit. A package groups modules. Python ships with the standard library; third-party packages are installed into an environment. Importing does not install a package.


```python
import math
from statistics import median
from collections import Counter
print(math.sqrt(81))
print(median([1, 100, 3]))
print(Counter("banana"))
```

## 2. A module that is also a script

The __name__ guard runs a demonstration only when the file is executed as a script. This example creates a module in a temporary folder and uses the current Python interpreter to run it. subprocess.run receives an argument list, without shell string construction.


```python
from pathlib import Path
from tempfile import TemporaryDirectory
import subprocess
import sys

module_source = 'def square(n):\n    return n * n\n\nif __name__ == "__main__":\n    print(square(7))\n'
with TemporaryDirectory() as folder:
    module_path = Path(folder) / "math_helpers.py"
    module_path.write_text(module_source, encoding="utf-8")
    result = subprocess.run([sys.executable, str(module_path)],
                            capture_output=True, text=True, check=True)
    print(result.stdout.strip())
assert result.stdout.strip() == "49"
```

**Try a change:** Change the demonstration input to 8 and rerun.


## 3. Know which interpreter runs

An environment keeps a project's installed packages separate. The README contains PowerShell setup commands. Commands such as python -m pip install belong in a terminal, not an ordinary Python cell. In notebooks, %pip targets the current kernel, but you may need to restart it after installation.


```python
import sys
import importlib.metadata
print("Kernel interpreter:", sys.executable)
print("ipykernel version:", importlib.metadata.version("ipykernel"))
```
