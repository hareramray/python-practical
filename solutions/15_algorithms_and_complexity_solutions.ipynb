## 1. Count work before timing

Big-O describes how work grows with input size; it is not an exact stopwatch prediction. A list scan is O(n), while set membership is expected O(1) after building the set. Constructing the set costs O(n) time and additional memory.


```python
def linear_find(items, target):
    comparisons = 0
    for index, item in enumerate(items):
        comparisons += 1
        if item == target:
            return index, comparisons
    return -1, comparisons

print(linear_find(list(range(10)), 9))
print(linear_find(list(range(100)), 99))
```

## 2. Binary search

For sorted random-access data, binary search halves the remaining interval each step: O(log n) comparisons. Sorting first costs O(n log n), so it may not pay off for a single lookup. This version returns any matching index or -1.


```python
def binary_search(items, target):
    low, high = 0, len(items) - 1
    while low <= high:
        middle = (low + high) // 2
        if items[middle] == target:
            return middle
        if items[middle] < target:
            low = middle + 1
        else:
            high = middle - 1
    return -1

assert binary_search([2, 4, 6, 8], 6) == 2
assert binary_search([], 6) == -1
assert binary_search([2, 4], 3) == -1
print(binary_search([2, 4, 6, 8], 8))
```

## 3. Recursion and memoization

Recursive functions solve smaller versions of a problem and need a base case. Caching avoids recomputing the same input. This Fibonacci version has O(n) distinct subproblems, though recursion depth still limits large n. An iterative approach avoids recursive stack growth.


```python
from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci(n):
    if n < 0:
        raise ValueError("n must be nonnegative")
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(20))
print(fibonacci.cache_info())
assert fibonacci(10) == 55
```

### Exercise 1: Stable deduplication

Return a new list with duplicates removed while preserving first-seen order. Inputs contain hashable values. Aim for expected O(n) time.

<details>
<summary>Show a hint</summary>

Use a set for membership and a list for output order.

</details>


```python
def unique_in_order(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result
```

### Exercise 2: Iterative factorial

Compute n! for a nonnegative integer using a loop. Return 1 for n=0; raise ValueError for negative n.

<details>
<summary>Show a hint</summary>

Start the accumulator at 1, not 0.

</details>


```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be nonnegative")
    result = 1
    for value in range(2, n + 1):
        result *= value
    return result
```
