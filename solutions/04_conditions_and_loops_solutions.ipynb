## 1. Conditions

Indented blocks belong to the statement above them. Python convention uses four spaces. Comparisons produce booleans; `and`, `or`, and `not` combine conditions. Only the first matching branch runs.


```python
score = 72
if score >= 80:
    grade = "A"
elif score >= 60:
    grade = "B"
else:
    grade = "Keep practicing"
print(grade)
print(0 <= score <= 100)
```

## 2. for loops

A for loop visits each item. range(1, 6) yields 1 through 5. enumerate() supplies an index and value; it avoids manually maintaining a counter.


```python
total = 0
for number in range(1, 6):
    total += number
print(total)
for position, topic in enumerate(["strings", "lists", "loops"], start=1):
    print(position, topic)
assert total == 15
```

## 3. while, break, and continue

A while loop repeats while its condition is true. Update something that can make the condition false. break exits a loop; continue skips the rest of the current iteration.


```python
remaining = 3
while remaining > 0:
    print("Attempts left:", remaining)
    remaining -= 1
for value in [3, -1, 8, 0, 10]:
    if value < 0:
        continue
    if value == 0:
        break
    print("Accepted:", value)
```
