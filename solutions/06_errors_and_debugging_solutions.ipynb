## 1. Read the failure

Exceptions describe why an operation failed. In a traceback, start with the last line, then find the relevant line in your code. These demonstrations catch their expected errors so Run All can complete.


```python
try:
    int("three")
except ValueError as error:
    print(type(error).__name__ + ":", error)
try:
    [10, 20][5]
except IndexError as error:
    print(type(error).__name__ + ":", error)
```

## 2. Validate a contract

Raise a clear exception when an input violates your function's contract. Catch only errors you know how to handle. Do not silently swallow every exception; that can hide bugs.


```python
def divide_bill(total, people):
    if people <= 0:
        raise ValueError("people must be positive")
    return total / people

try:
    print(divide_bill(100, 0))
except ValueError as error:
    print("Please correct the input:", error)
assert divide_bill(100, 4) == 25
```

## 3. Debug a logic error

A logic bug can return the wrong value without raising an exception. Reduce the problem to a tiny input with a known answer. Inspect intermediate values before changing code.


```python
def buggy_average(values):
    return sum(values) / (len(values) + 1)

sample = [10, 20]
print("Observed:", buggy_average(sample), "Expected:", 15)
print("Sum:", sum(sample), "Count:", len(sample))
# In an interactive session, uncomment breakpoint() to inspect state.
# breakpoint()
```

### Exercise 2: Correct average

Return the arithmetic mean; raise ValueError for an empty list.

<details>
<summary>Show a hint</summary>

Check emptiness before dividing.

</details>


```python
def average(values):
    if not values:
        raise ValueError("values must not be empty")
    return sum(values) / len(values)
```
