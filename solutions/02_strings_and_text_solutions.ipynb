## 1. Characters and slices

A string is an immutable sequence of characters. Indexes start at 0; negative indexes count from the end. A slice includes its start but excludes its stop. Slices create new strings.


```python
language = "Python"
print(language[0], language[-1])
print(language[:3], language[3:], language[::-1])
print(len(language))
```

## 2. Clean input

Methods are operations accessed with a dot. String methods return new strings, so keep the returned value. `strip()` removes surrounding whitespace, not spaces inside the text.


```python
raw = "  Learn Python  "
cleaned = raw.strip().lower()
print(repr(raw), repr(cleaned))
print(cleaned.replace("python", "coding"))
print("python" in cleaned, cleaned.startswith("learn"))
```

**Try a change:** Try tabs or a newline at each end of raw. Use repr() to see them.


## 3. Parse and format

`split()` breaks text into a list of parts. `join()` combines strings using a separator. Real CSV can contain quoted commas; use the csv module in notebook 07 for that.


```python
items = "tea,coffee,milk".split(",")
print(items)
print(" | ".join(items))
name, score = "Mira", 92.456
print(f"{name} scored {score:.1f}%")
```
