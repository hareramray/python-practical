## 1. Lists and tuples

A list is an ordered, mutable sequence. A tuple is an ordered sequence whose references cannot be reassigned. Use tuples for fixed groups such as coordinates. A tuple may still contain mutable objects.


```python
tasks = ["read", "code"]
tasks.append("review")
tasks[0] = "predict"
print(tasks, tasks[:2])
print("Removed:", tasks.pop())
point = (10, 20)
x, y = point
print(x, y)
```

## 2. Dictionaries and sets

A dictionary maps unique keys to values. `get()` supplies a fallback for a missing key. A set stores unique hashable values and supports membership and set operations; do not depend on its display order.


```python
student = {"name": "Mira", "score": 80}
student["score"] = 85
print(student["name"], student.get("city", "Unknown"))
topics = {"lists", "strings", "lists"}
print(sorted(topics))
print(topics & {"strings", "loops"})
```

## 3. Sharing versus copying

Assigning a list to another name does not copy it. Both names refer to the same object. `copy()` makes a shallow copy: nested objects remain shared.


```python
original = [1, 2]
alias = original
copied = original.copy()
alias.append(3)
print(original, copied)
assert alias is original
nested = [[1], [2]]
shallow = nested.copy()
shallow[0].append(9)
print(nested)
```
