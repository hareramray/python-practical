## 1. Input, work, output

A function groups reusable behavior. Parameters name inputs; return sends a result to the caller. print displays a value but does not return it. A function without return returns None.


```python
def rectangle_area(width, height):
    """Return area in square units."""
    return width * height

area = rectangle_area(4, 3)
print(area)
assert rectangle_area(0, 3) == 0
```

## 2. Defaults and keyword arguments

Defaults let callers omit some arguments. A * in a signature makes later parameters keyword-only. Local names inside a function normally stay inside it; prefer explicit inputs over reliance on global variables.


```python
def final_price(price, *, discount=0.0):
    return price * (1 - discount)

print(final_price(100))
print(final_price(100, discount=0.2))
def add_all(*numbers):
    return sum(numbers)
print(add_all(1, 2, 3))
```

**Try a change:** Call add_all() with no arguments. Predict its result.


## 3. Safe mutable defaults

Default arguments are evaluated once when the function is defined. A default list would be shared across calls. Use None and create a new list inside when you need independent containers.


```python
def add_task(task, tasks=None):
    if tasks is None:
        tasks = []
    tasks.append(task)
    return tasks

print(add_task("read"))
print(add_task("code"))
assert add_task("a") == ["a"]
```

**Try a change:** Pass an existing list as tasks. Observe that this function mutates it.
