## 1. Closures

A closure is a function that remembers names from an enclosing scope. Functions are objects, so they can be returned and passed around. This creates configured behavior without global variables.


```python
def make_multiplier(factor):
    def multiply(value):
        return value * factor
    return multiply

double = make_multiplier(2)
triple = make_multiplier(3)
print(double(5), triple(5))
```

## 2. Decorators

A decorator receives a function and returns a replacement. @trace is shorthand for add = trace(add). *args and **kwargs forward positional and keyword arguments; wraps preserves metadata for documentation and debugging.


```python
from functools import wraps

def trace(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print("Calling", function.__name__)
        return function(*args, **kwargs)
    return wrapper

@trace
def add(a, b):
    """Add two values."""
    return a + b

assert add(2, b=3) == 5
print(add.__name__, add.__doc__)
```

**Try a change:** Decorate another function. Verify that its return value is preserved.


## 3. Cleanup even on failure

A context manager controls entry and exit around a with block. In a contextmanager generator, code before yield acquires a resource and finally releases it. finally runs even if the block raises.


```python
from contextlib import contextmanager

@contextmanager
def recorded_session(events):
    events.append("open")
    try:
        yield events
    finally:
        events.append("close")

events = []
try:
    with recorded_session(events):
        events.append("work")
        raise ValueError("demonstration")
except ValueError:
    pass
print(events)
assert events == ["open", "work", "close"]
```
