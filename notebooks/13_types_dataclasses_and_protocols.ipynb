## 1. Hints describe a contract

Annotations help readers and static analyzers. Python does not enforce them automatically. Use explicit checks for runtime validation. The union syntax X | None represents an optional value.


```python
def first_long_word(words: list[str], minimum: int = 5) -> str | None:
    for word in words:
        if len(word) >= minimum:
            return word
    return None

print(first_long_word(["a", "python", "code"]))
assert first_long_word([]) is None
print(first_long_word.__annotations__)
```

## 2. Dataclasses

A dataclass generates initialization, representation, and equality methods. frozen=True prevents ordinary attribute reassignment; it does not deeply freeze nested data. default_factory creates a fresh mutable container for each instance.


```python
from dataclasses import dataclass, field

@dataclass(frozen=True)
class Product:
    name: str
    price_minor: int

    def __post_init__(self):
        if self.price_minor < 0:
            raise ValueError("price cannot be negative")

@dataclass
class Basket:
    products: list[Product] = field(default_factory=list)

a, b = Basket(), Basket()
a.products.append(Product("Pen", 1000))
print(a, b)
assert b.products == []
```

## 3. Structural interfaces

A Protocol describes operations a consumer needs. Implementations need not inherit from it. A static type checker can compare a class to this interface; these annotations alone do not run such a checker.


```python
from typing import Protocol

class Sender(Protocol):
    def send(self, message: str) -> str: ...

class PreviewSender:
    def send(self, message: str) -> str:
        return f"Preview: {message}"

def notify(sender: Sender, message: str) -> str:
    return sender.send(message)

print(notify(PreviewSender(), "Practice complete"))
```
