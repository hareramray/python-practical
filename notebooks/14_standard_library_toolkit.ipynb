## 1. Dates and aware times

date models calendar dates; timedelta models durations. Use timezone-aware datetimes for instants. A fixed +05:30 offset below needs no timezone database; geographic timezones with daylight-saving changes need zoneinfo and appropriate timezone data.


```python
from datetime import date, datetime, timedelta, timezone
start = date(2025, 1, 1)
print("One week later:", start + timedelta(days=7))
utc_time = datetime(2025, 1, 1, 12, tzinfo=timezone.utc)
india_offset = timezone(timedelta(hours=5, minutes=30))
print(utc_time.astimezone(india_offset).isoformat())
```

## 2. Decimal from text

Decimal constructed from a string preserves the intended decimal value. Constructing it from a float also preserves that float's approximation. Choose a rounding rule explicitly when reducing precision.


```python
from decimal import Decimal, ROUND_HALF_UP
price = Decimal("19.99")
total = (price * 3 * Decimal("1.05")).quantize(Decimal("0.01"), rounding=ROUND_HALF_UP)
print(total)
assert Decimal("0.1") + Decimal("0.2") == Decimal("0.3")
assert total == Decimal("62.97")
```

## 3. Focused patterns and queues

Regular expressions match text patterns. Use raw strings for patterns with backslashes. fullmatch requires the entire string to fit. A deque efficiently adds and removes items at both ends.


```python
import re
from collections import deque, defaultdict
pattern = re.compile(r"TASK-[0-9]{3}")
print(bool(pattern.fullmatch("TASK-042")))
print(bool(pattern.fullmatch("xTASK-042")))
queue = deque(["read", "code"])
queue.append("review")
print(queue.popleft(), list(queue))
grouped = defaultdict(list)
for category, item in [("food", "rice"), ("tools", "pen"), ("food", "tea")]:
    grouped[category].append(item)
print(dict(grouped))
```
