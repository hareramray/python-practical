## 1. Paths and text files

Path represents a filesystem path. Use explicit UTF-8 encoding for portable text. We use temporary directories so running examples never overwrites your own documents. Files disappear when the with block exits.


```python
from pathlib import Path
from tempfile import TemporaryDirectory

with TemporaryDirectory() as folder:
    path = Path(folder) / "notes.txt"
    path.write_text("Learn\nPractice\nReview\n", encoding="utf-8")
    print(path.read_text(encoding="utf-8"))
    with path.open(encoding="utf-8") as handle:
        lines = [line.strip() for line in handle]
assert lines == ["Learn", "Practice", "Review"]
```

## 2. JSON

JSON stores objects, arrays, text, numbers, booleans, and null. Python dictionaries and lists map naturally to it. dumps/loads work on strings; dump/load work with file handles. Dates and sets need explicit conversion.


```python
import json
profile = {"name": "Mira", "topics": ["files", "json"], "active": True}
encoded = json.dumps(profile, indent=2)
print(encoded)
restored = json.loads(encoded)
assert restored == profile
```

## 3. CSV

CSV is tabular text. DictReader uses the header row as dictionary keys. Values arrive as strings and need conversion. Use newline='' when opening CSV files so the csv module controls line endings.


```python
import csv
import io
source = io.StringIO('item,quantity,price\npen,2,10\n"book, ruled",1,80\n')
rows = list(csv.DictReader(source))
total = sum(int(row["quantity"]) * int(row["price"]) for row in rows)
print(rows)
print("Total:", total)
assert total == 100
```
