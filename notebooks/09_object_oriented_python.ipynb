## 1. Define a class

A class describes objects with related data and behavior. __init__ initializes each instance. self refers to that instance. Each BankAccount below has its own balance, represented in integer minor units to avoid float rounding.


```python
class BankAccount:
    def __init__(self, owner, balance=0):
        if balance < 0:
            raise ValueError("balance cannot be negative")
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("deposit must be positive")
        self._balance += amount

    def __repr__(self):
        return f"BankAccount({self.owner!r}, balance={self.balance})"

account = BankAccount("Mira", 1000)
account.deposit(500)
print(account)
assert account.balance == 1500
```

## 2. Compose objects

Composition means an object uses another object. This often keeps responsibilities clearer than a deep inheritance tree. A Wallet delegates storage and validation to an account instead of reimplementing it.


```python
class Wallet:
    def __init__(self, account):
        self.account = account

    def summary(self):
        return f"{self.account.owner}: {self.account.balance} minor units"

wallet = Wallet(account)
print(wallet.summary())
```

**Try a change:** Pass a different account to a new wallet. Explain which object owns the balance.


## 3. Inherit only for a useful relationship

Inheritance can specialize behavior when the new type can stand in for the original. Python also supports duck typing: code can work with any object providing the operations it needs, regardless of inheritance.


```python
class Animal:
    def speak(self):
        return "sound"

class Dog(Animal):
    def speak(self):
        return "woof"

def announce(animal):
    return animal.speak().upper()

print(announce(Dog()))
assert isinstance(Dog(), Animal)
```
