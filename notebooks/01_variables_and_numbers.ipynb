## 1. Values have types

A variable is a name bound to a value. Integers represent whole numbers, floats approximate decimal numbers, booleans are True/False, and None means no value. `type()` reveals a value's type.


```python
age = 20
price = 49.5
is_learning = True
result = None
print(type(age), type(price), type(is_learning), type(result))
age = age + 1
print("Next age:", age)
```

## 2. Arithmetic

`/` is division, `//` is floor division, `%` gives the remainder, and `**` raises a power. Parentheses control evaluation order. Floor division rounds down, including for negative values.


```python
print(17 / 5, 17 // 5, 17 % 5)
print(2 ** 3, (2 + 3) * 4)
print(-7 // 3)
minutes = 135
print("Hours:", minutes // 60, "Remaining minutes:", minutes % 60)
```

**Try a change:** Convert 200 minutes into hours and remaining minutes.


## 3. Conversion and approximate decimals

`int()` and `float()` convert numeric text. An f-string inserts values into text. Many decimal fractions cannot be represented exactly as binary floats. Use a tolerance when comparing calculated floats; use Decimal for money later.


```python
raw_quantity = "3"
total = int(raw_quantity) * 19.5
print(f"Total: {total:.2f}")
print(0.1 + 0.2)
import math
assert math.isclose(0.1 + 0.2, 0.3)
```
