## 1. Comprehensions

A comprehension is a compact transformation, optionally with a filter. Prefer an ordinary loop if nesting or side effects make the expression hard to read.


```python
numbers = range(1, 7)
squares = [n * n for n in numbers if n % 2 == 0]
lengths = {word: len(word) for word in ["code", "practice"]}
print(squares, lengths)
assert squares == [4, 16, 36]
```

## 2. Iterator protocol

An iterable can produce an iterator. next() consumes one item; an exhausted iterator stays exhausted. The optional default avoids StopIteration. A list is reusable; a particular iterator over it is consumed.


```python
values = [10, 20]
iterator = iter(values)
print(next(iterator), next(iterator))
print(next(iterator, "finished"))
print(list(iterator), list(values))
```

## 3. yield and streaming

A generator function pauses at yield and resumes on the next request. Lazy pipelines avoid materializing all intermediate values. They still do work when consumed, and are normally single-use.


```python
def running_totals(values):
    total = 0
    for value in values:
        total += value
        yield total

totals = running_totals([3, 4, 5])
print(next(totals))
print(list(totals))
print(sum(n * n for n in range(100)))
assert list(running_totals([])) == []
```
