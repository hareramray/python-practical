## 1. Arrays and vectorization

NumPy arrays store values of a common dtype and support elementwise operations. Multiplying a Python list repeats it; multiplying an array multiplies its elements. shape gives the length of each dimension.


```python
import numpy as np
prices = np.array([10.0, 20.0, 30.0])
print("List:", [1, 2] * 2)
print("Array:", np.array([1, 2]) * 2)
print(prices.shape, prices.dtype)
discounted = prices * 0.9
print(discounted)
np.testing.assert_allclose(discounted, [9, 18, 27])
```

## 2. Axes and broadcasting

For a two-dimensional array, axis=0 reduces over rows, leaving one value per column; axis=1 reduces over columns. Broadcasting aligns trailing dimensions when sizes match or one size is 1. @ performs matrix multiplication, while * multiplies elementwise.


```python
scores = np.array([[70, 80, 90], [60, 75, 85]])
print("Shape:", scores.shape)
print("Subject means:", scores.mean(axis=0))
print("Student means:", scores.mean(axis=1))
bonus = np.array([5, 0, 2])
print("Adjusted:\n", scores + bonus)
weights = np.array([0.2, 0.3, 0.5])
print("Weighted scores:", scores @ weights)
```

## 3. Masks, views, and copies

A boolean mask selects matching values. Basic slicing often produces a view sharing memory with the original. Use copy() when you need independent storage. Boolean indexing produces a new array.


```python
values = np.array([1, 2, 3, 4])
view = values[1:3]
view[0] = 99
print(values)
independent = values.copy()
independent[0] = -1
print("Selected:", values[values > 3])
assert values[0] == 1
assert np.shares_memory(values, view)
```
