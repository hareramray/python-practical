## 1. A function with a contract

A good test checks observable behavior against an independently known result. Include boundaries and invalid inputs. Keep production validation explicit, and use assertions in tests.


```python
def shipping_cost(total):
    """Return shipping in whole currency units; free at 500 or above."""
    if total < 0:
        raise ValueError("total cannot be negative")
    return 0 if total >= 500 else 40

assert shipping_cost(0) == 40
assert shipping_cost(499) == 40
assert shipping_cost(500) == 0
```

## 2. A small test suite

unittest is built into Python. Each test method describes one behavior. The explicit runner works inside notebooks without unittest.main() consuming Jupyter's command-line arguments.


```python
import unittest

class ShippingTests(unittest.TestCase):
    def test_threshold(self):
        self.assertEqual(shipping_cost(500), 0)

    def test_below_threshold(self):
        self.assertEqual(shipping_cost(499), 40)

    def test_negative_rejected(self):
        with self.assertRaises(ValueError):
            shipping_cost(-1)

suite = unittest.defaultTestLoader.loadTestsFromTestCase(ShippingTests)
result = unittest.TextTestRunner(verbosity=2).run(suite)
assert result.wasSuccessful()
```

**Try a change:** Temporarily introduce the >= to > bug, run the suite, then restore it.


## 3. Test changing external inputs

Passing a dependency as an argument makes a function easier to test. Here the caller supplies a clock, so a test never depends on the actual day. Similar techniques work for HTTP clients and file readers.


```python
from datetime import date

def greeting_for_today(name, today):
    return f"Hello {name}, today is {today().isoformat()}"

def fixed_day():
    return date(2025, 1, 2)

assert greeting_for_today("Mira", fixed_day) == "Hello Mira, today is 2025-01-02"
print(greeting_for_today("Mira", fixed_day))
```
